```python
# Student grade A - above 90 marks, B - above 80 marks, C - above 70 marks, D - above 60 marks, F - below 60 marks
# if elif else format
marks = int(input("Enter your marks: "))
if marks > 90:
   print("Your grade is: A")
elif marks > 80:
    print("Your grade is: B")
elif marks > 70:
    print("Your grade is: C")
elif marks > 60:
    print("Your grade is: D")
else:
    print("Your grade is: F")
```
