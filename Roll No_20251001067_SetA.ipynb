Q1) Write a simple python program for equilibrium reaction A + B gives C. Where Kcx^2-(2Kc+1)x+Kc=0. It must ask Kc as input. Show clear message if Kc is not a number or if Kc<=0. A is 1-x B is 1-x C is x. Find values of A, B, C. Calculate discriminant and roots of quadratic equation. Select meaningful root 0<x<1. Print A B C upto 3 decimals. Find conversion of limiting reactant using =x/A*100 in percent. Classify conversion using if statement. Conversion >=70 "high", >=40 "moderate" otherwise low Do it without importing math.

```python
Kc = input("Enter the equilibrium constant Kc: ")
print("Initial datatype of Kc:", type(Kc))
[Ao, Bo, Co] = [1, 1, 0]

try:
    Kc = float(Kc)
except ValueError:
    print("Invalid input: Kc must be a number.")
else:
    print("Kc=", Kc)
    
    if Kc <= 0:
        print("Invalid input: Kc must be greater than 0.")
    else:
        print("Converted Datatype of Kc:", type(Kc))
        coefficient_a = Kc
        coefficient_b = -(2 * Kc + 1)
        coefficient_c = Kc
        print(f"a = {coefficient_a}")
        print(f"b = {coefficient_b}")
        print(f"c = {coefficient_c}")

        discriminant = coefficient_b ** 2 - 4 * coefficient_a * coefficient_c
        square_root = discriminant ** 0.5

        root_1 = (-coefficient_b + square_root) / (2 * coefficient_a)
        root_2 = (-coefficient_b - square_root) / (2 * coefficient_a)

        print(f"Discriminant: {discriminant:.3f}")
        print(f"Root 1: {root_1:.6f}")
        print(f"Root 2: {root_2:.6f}")

        x = None
        for root in (root_1, root_2):
            if 0 < root < 1:
                x = root
                break

        if x is None:
            print("No meaningful root found in the range 0 < x < 1.")
        else:
            print(f"Meaningful root: {x:.6f}")
            A = 1 - x
            B = 1 - x
            C = x
            conversion = (x / Ao) * 100

            print(f"\nEquilibrium concentrations:")
            print(f"[A] = {A:.3f}")
            print(f"[B] = {B:.3f}")
            print(f"[C] = {C:.3f}")
            print(f"Limiting reactant conversion = {conversion:.3f}%")

            if conversion >= 70:
                classification = "high"
            elif conversion >= 40:
                classification = "moderate"
            else:
                classification = "low"

            print(f"Conversion classification: {classification}")
```

Q2) A)Develop a simple program  For loop for equation dC/dt= -kC. C_new=C_old- kCdt. Co is 2.0 mol/L k is 0.08 min^-1 and dt is 1min. Simulate 10 min and print table of Time(min), Concentration(mol/L), and Conversion(%). Use math.exp(), calculate exact Coe^(-kt) at 10 min and print absolute error of previous result. Use while loop to repeat untill conversion reaches 60% or time reaches 100 min. Print time needed and final concentration

```python
import math

Co = 2.0
k = 0.08
dt = 1.0
simulation_time = 10

print("Euler simulation for dC/dt = -kC")
print(f"{'Time (min)':>10} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")

C_old = Co
print(f"{0:10.1f} {C_old:24.4f} {0.0:18.2f}")

for step in range(1, simulation_time + 1):
    C_new = C_old - k * C_old * dt
    conversion_percent = (Co - C_new) / Co * 100
    print(f"{step * dt:10.1f} {C_new:24.4f} {conversion_percent:18.2f}")
    C_old = C_new

exact_concentration = Co * math.exp(-k * simulation_time)
absolute_error = abs(C_old - exact_concentration)

print(f"\nExact concentration at {simulation_time} min: {exact_concentration:.6f} mol/L")
print(f"Absolute error at {simulation_time} min: {absolute_error:.6f} mol/L")

C_old = Co
time_min = 0.0
conversion_percent = 0.0

while conversion_percent < 60 and time_min < 100:
    C_new = C_old - k * C_old * dt
    C_old = C_new
    time_min += dt
    conversion_percent = (Co - C_old) / Co * 100

if conversion_percent >= 60:
    print("\nThe 60% conversion target was reached.")
else:
    print("\nThe 100 min time limit was reached before 60% conversion.")

print(f"Time needed: {time_min:.1f} min")
print(f"Final concentration: {C_old:.4f} mol/L")
```

Q2) B)Batch labels such as PLANT20261030 are stored as text. Write a simple program that reads a label, loops through each, character prints the number of digits an sum of digits using isdigit. Show a message for empty label and label with no digits.

```python
batch_label = input("Enter a batch label: ").strip()

if batch_label == "":
    print("The batch label is empty.")
else:
    print("The batch label is:", batch_label)
    digit_count = 0
    digit_sum = 0

    for character in batch_label:
        if character.isdigit():
            digit_count += 1
            digit_sum += int(character)

    print(f"Number of digits: {digit_count}")
    print(f"Sum of digits: {digit_sum}")

    if digit_count == 0:
        print("The batch label contains no digits.")
```

Q3)Prepare a simple program for testing n candidate additives, r of them together (order does not matter) in one formulation. Write recursive function factorial (n) add base case zero factorial is 1. Function Combination(n,r)=n!/r!(n-r)!. Reject invalid input negative values, r>n.

```python
def factorial(number):
    if number == 0:
        return 1
    return number * factorial(number - 1)


def Combination(n, r):
    return factorial(n) // (factorial(r) * factorial(n - r))


try:
    n = int(input("Enter the total number of candidate additives (n): "))
    r = int(input("Enter the number of additives in one formulation (r): "))
except ValueError:
    print("Invalid input: n and r must be whole numbers.")
else:
    if n < 0 or r < 0:
        print("Invalid input: n and r cannot be negative.")
    elif r > n:
        print("Invalid input: r cannot be greater than n.")
    else:
        print(f"n = {n}, r = {r}")
        number_of_formulations = Combination(n, r)
        print(f"Number of possible formulations: {number_of_formulations}")
```

Q3) B)Write a simple program for reaction whose sample time is = Fibonacci number*2 min. Write an iterative function fib_iter(n) that returns a list of first n Fibonacci numbers and recursive function fib_rec(k) that returns the kth Fibonacci number. Let Fo=0, F1=1. For n=10 check both functions. Print a table: Sample No. | Fibonacci tag| Sampling time(min)

```python
def fib_iter(n):
    if n < 0:
        raise ValueError("n cannot be negative.")

    fibonacci_numbers = []
    previous_number = 0
    current_number = 1

    for _ in range(n):
        fibonacci_numbers.append(previous_number)
        previous_number, current_number = current_number, previous_number + current_number

    return fibonacci_numbers


def fib_rec(k):
    if k < 0:
        raise ValueError("k cannot be negative.")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = 10
fibonacci_numbers = fib_iter(n)

print(f"Iterative Fibonacci numbers for n = {n}: {fibonacci_numbers}")
print(f"Recursive Fibonacci numbers for n = {n}: {[fib_rec(i) for i in range(n)]}")
print("Recursive = Iterative check:", all(fib_rec(index) == value for index, value in enumerate(fibonacci_numbers)))
print()
print(f"{'Sample No.':<12}{'Fibonacci tag':<18}{'Sampling time (min)':>20}")
print("-" * 50)

for sample_number, fibonacci_tag in enumerate(fibonacci_numbers, start=1):
    fibonacci_index = sample_number - 1
    sampling_time = fibonacci_tag * 2
    print(f"{sample_number:<12}F{fibonacci_index} = {fibonacci_tag:<13}{sampling_time:>16}")
```
